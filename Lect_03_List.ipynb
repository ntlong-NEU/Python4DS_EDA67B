```python
alist = [1,2]
print(alist)
blist = alist
blist.append('new')
print(alist)
print(blist)
```

```python
alist = [1,2]
print(alist)
blist = alist[:] #slicing as copy
blist.append('new')
print(alist)
print(blist)
```

```python
# shalow copy and deep copy
alist = [1,[2,1]]
print(alist)
blist = alist[:] #slicing as copy
blist[1][1] = 'new'
print(alist)
print(blist)
```
