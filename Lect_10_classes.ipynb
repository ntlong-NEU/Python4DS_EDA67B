```python
class Dog:
    '''docstring of the class'''
    def __init__(self, name, age):
        "docstring of method"
        self.name = name
        self.age = age
    def sit(self):
        print(f'{self.name} is sitting')
    def get_info(self):
        return {"name":self.name, 'age':self.age}
```

```python
my_dog = Dog('Xoai', 3)
your_dog = Dog('Mic', 5)
```

```python
class Cat:
    def __init__(self, name, age):
        self.name = name
        self._age = age
        self.__id = 123120282
    def get_info(self):
        return {"name":self.name, "age":self._age}
    def get_id(self):
        print(self.__id)
```

```python
class Stack:
    def __init__(self):
        self.__stack = []
    def push(self, item):
        self.__stack.append(item)
    def pop(self):
        if not self.is_empty():
            return self.__stack.pop()
        else:
            print('The stack is empty')
    def is_empty(self):
        return not bool(self.__stack)
    def peek(self):
        if not self.is_empty():
            return self.__stack[-1]
        else:
            print('The stack is empty')
    def size(self):
        return len(self.__stack)
    def __str__(self):
        return str(self.__stack)
```

```python
my_stack = Stack()
```

```python
my_stack.size()
```

```python
my_stack.pop()
```

```python
my_stack.peek()
```

```python
my_stack.push(1)
my_stack.push(2)
my_stack.push(3)
```

```python
print(my_stack)
```

```python
my_stack.pop()
print(my_stack)
```

```python
print(my_stack.peek())
print(my_stack)
```
