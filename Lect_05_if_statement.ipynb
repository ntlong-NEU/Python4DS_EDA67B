```python
text = '''ChatGPT is    an AI assistant.\n\n It understands natural language.\n\n It helps answer    questions fast.\n\n It can assist with writing \n\ntasks.\n\n ChatGPT learns    from vast data.\n\nIt adapts to    user needs well.\n\n This AI     is by OpenAI's team.\n\n It's built to aid     and inform.\n\n Ask ChatGPT anything today!'''
```

```python
text = text.replace('\n',' ')
text = text.replace('  ',' ')
text 
```

```python
text.index('.')
```

```python
endpoint = [i for i in range(len(text)) if text[i] in {'.','?','!'}]
#endpoint.insert(0,0)
endpoint
```

```python
max_chars = 61
```

```python
cut_points = []
start = 0
end = len(text)
for i in range(len(endpoint)):
    p = endpoint[i]
    if p- start >= max_chars:
        end  = endpoint[i-1]+1
        cut_points.append((start,end))
        start = end +1
cut_points.append((start,len(text)))
```

```python
cut_points
```

```python
for s,e in cut_points:
    print(text[s:e])
```
