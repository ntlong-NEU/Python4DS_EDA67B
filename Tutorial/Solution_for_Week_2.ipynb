The expected value (mean) is the most fundamental statistical metric. Given a list of web traffic over 5 days: `traffic = [1200, 1500, 900, 2000, 1300]`  
**Requirement:** Using `for loop` to calculate and then print the mean of the `traffic` array.

**Question 5:**  
A coffee shop wants to organize daily sale data into a clean, easy-to-read table. Here their data:
```
item_names = ["Flat White", "Capuchino", "Americano", "Long Black", "Latte"]
prices = [6, 7, 5.5, 5, 6.5]
quantities = [10, 14, 18, 12, 15]
```
1. Use a List Comprehension to calculate the total cost for each item based on the three lists provided above, and store these values in a new list `total_cost`

2. Present this data as a professional-looking receipt on the console. Your receipt must meet the following formatting criteria:

- Use string multiplication to create a top and bottom border (each consisting of exactly 45 = signs).

- Include a header row with the columns: `Item Name, Price, Qty, and Total`. Immediately below the header, add a separator line made of exactly 45 - signs.

- Use f-string alignment features to ensure all columns are neatly vertically aligned.

- Ensure that all money-related columns (`Price and Total`) consistently display exactly 2 decimal places.

````python
item_names = ["Flat White", "Capuchino", "Americano", "Long Black", "Latte"]
prices = [6, 7, 5.5, 5, 6.5]
quantities = [10, 14, 18, 12, 15]
total=[prices[i]*quantities[i] for i in range(len(item_names))]
print("="*45)
print(f"{'Item Name':^14}{'Price':^10}{'Qty':^8}{'Total':^13}")
print("-"*45)
for j in range(len(item_names)):
  print(f"{item_names[j]:^14}{prices[j]:^10.2f}{quantities[j]:^8}{total[j]:^13.2f}")
print("="*45)
````

````python
#Zip
item_names = ["Flat White", "Capuchino", "Americano", "Long Black", "Latte"]
prices = [6, 7, 5.5, 5, 6.5]
quantities = [10, 14, 18, 12, 15]
total = [p*q for p,q in zip(prices,quantities)]
print("="*45)
print(f"{'Item Name':^14}{'Price':^10}{'Qty':^8}{'Total':^13}")
print("-"*45)
for i in range(len(item_names)):
  print(f"{item_names[i]:^14}{prices[i]:^10.2f}{quantities[i]:^8}{total[i]:^13.2f}")
print("="*45)
````

**Question 6:**  
Machine Learning algorithms often require scaling data to a range between $[0, 1]$. The scaling formula for a value $x$ is:
$$x_{scaled} = \frac{x - min}{max - min}$$
Given a list of raw scores:
`raw_scores = [10, 50, 20, 90, 30]`  
**Requirement:**
1. Write a code block (without using built-in `min`/`max` functions) to find the `max` and `min` values in the array.
2. Write a loop to calculate and store the $x_{scaled}$ values into a new list named `scaled_scores`.



````python
#Write your code here
raw_scores = [10, 50, 20, 90, 30]
sorted_scores = sorted(raw_scores)
print("Min: ", sorted_scores[0])
print("Max: ", sorted_scores[-1])
scaled_scores = []
for i in sorted_scores:
  scaled = ((i - sorted_scores[0]) / (sorted_scores[-1] - sorted_scores[0]))
  scaled_scores.append(scaled)
print(scaled_scores)
````

**Question 7:**   
Given two matrices $P$ and $Q$:
```
P = [[2, 4], [6, 8]]  
Q = [[0.5, 0.25], [2, 0.5]]
```
How to multiply matrix: [Link here](https://en.wikipedia.org/wiki/Matrix_multiplication)

**Requirement:** Initialize a zero-filled matrix $M$ of the appropriate dimensions. Using nested for loops, calculate the matrix product of $P$ and $Q$ according to the rules of matrix multiplication (multiplying the rows of $P$ by the columns of $Q$).

Print the resulting matrix $M$.

````python
#Write your code here
P = [[2, 4], [6, 8]]
Q = [[0.5, 0.25], [2, 0.5]]
M=[[0,0],[0,0]]
for i in range(len(P)):
  for j in range(len(Q[0])):
    for k in range(len(Q)):
      M[i][j]+=P[i][k]*Q[k][j]
print(M)
````
