**Question 7**

Create a Class MemoryCard

- **Attributes:**
  - `capacity` — the total storage capacity of the card.
  - `used_space` — the amount of space currently used, = 0 by default.

- **Methods:**
  - `save_file(file_size)` — calculates the remaining space (capacity - used_space). If the remaining space is greater than or equal to `file_size`, it adds `file_size` to `used_space` and prints "File saved successfully". If not, it prints "Error: Memory card is full!".
  - `format_card()` — resets the `used_space` to 0 and prints "Memory card formatted. All data cleared.".

Create a Class DigitalCamera

- **Attributes:**
  - `brand` — the brand of the camera.
  - `resolution` — the megapixel count, which determines how large the photo files will be.
  - `storage` — this attribute must be an instance of the MemoryCard class, passed in during initialization.

- **Methods:**
  - `take_photo()` — calculates the `photo_size` (formula: resolution * 2 MB). Then, it accesses the storage attribute and calls its `save_file(photo_size)` method. If the file is saved successfully, print "Snap! Photo taken by [brand] camera."
  - `check_remaining_space()` — accesses the memory card to calculate and print the exact amount of free space left.

**Scenario:**
- Create a MemoryCard instance with a small capacity of 50 MB.
- Create a DigitalCamera instance (Brand: "Sony", Resolution: 12 megapixels), assigning the memory card object to its storage attribute.
- Call `take_photo()` unitl the memory card is full.
- Call `format_card()` method directly through the camera's storage attribute to clear the memory.
- Call `take_photo()` once more to prove that the camera can take pictures again after formatting.
