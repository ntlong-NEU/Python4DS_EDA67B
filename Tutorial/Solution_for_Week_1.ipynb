**Question 1**

- Create a variable `student_name` storing the string `"Alice"`.

- Create a variable `gpa` storing the float `3.85`.

- Create a variable `is_enrolled` storing the boolean `True`.

- Cast `gpa` into an integer and save it in a new variable `gpa_int`.

- Cast `is_enrolled` into a string and save it in a new variable `enroll_str`.

- Use a single `print()` to output `student_name, gpa_int, and enroll_str` separated by a dash (-) using the sep parameter.

```python
student_name = "Alice"
gpa = 3.85
is_enrolled = True
gpa_int = int(gpa)
enroll_str = str(is_enrolled)
print(student_name, gpa_int, enroll_str, sep="-")
```

**Question 2**  
Given: `text = "  Learning Python is FUN!  "`

- Remove the extra spaces at the beginning and end, and save it to `clean_text`.

- Print `clean_text` in all lowercase letters.

- Print `clean_text` in Title Case (first letter of each word capitalized).

- Print a boolean checking if `clean_text` starts with `"Learning"`.

- Replace `"FUN"` with `"AWESOME"` in `clean_text` and print the result.

- Print a boolean checking if `clean_text` ends with the character `"!"`.

- Print the total length (number of characters) of `clean_text`.

```python
text = " Learning Python is FUN! "
clean_text = text.strip()
print(clean_text.lower())
print(clean_text.title())
print(clean_text.startswith("Learning"))
print(clean_text.upper().replace("FUN","AWESOME"))
print(clean_text.endswith("!"))
print(len(clean_text))
```

**Question 3**

- Prompt the user with `"Enter first float: "` and store it as a float `num1`.

- Prompt the user with `"Enter second float: " `and store it as a float `num2`.

- Calculate and print the addition of `num1` and `num2`.

- Calculate and print the floor division (//) of `num1` by `num2`.

- Calculate and print the remainder (modulo) when `num1` is divided by `num2`.

- Calculate `num1` divided by `num2`, round the result to exactly 3 decimal places using the round() function, and print it.

```python
num1 = float(input("Enter first float: "))
num2 = float(input("Enter second float: "))
print(num1 + num2)
print(num1 // num2)
print(num1 % num2)
print(round(num1/num2, 3))
```

**Question 4**

- Prompt the user for a `city` (string).

- Prompt the user for a `temperature` (float).

- Print `"City: [city]"` using an f-string, ensuring the city name is printed in UPPERCASE.

- Print` "Temp: [temperature]"` using an f-string, ensuring the temperature shows exactly 1 decimal place.

- Without using `.replace()`, print the `city` 3 times on the same line, separated by `"***"` (Hint: Use print() parameters).

```python
city = input("Enter a city's name: ")
temperature = float(input("Enter temperature: "))
print(f"City: {city.upper()}")
print(f"Temp: {temperature:.1f}")
print(city, city, city, sep = "***")
```

**Question 5**

- Prompt the user to input an `item_name`.

- Prompt the user to input a `price` as `float`.

- Prompt the user to input a `quantity` as `integer`.

- Calculate the `total_cost = price * quantity`.

- Print a top border using exactly 45 equal signs (=) using string multiplication.

- Print a header row: `"Item Name", "Price", "Qty", and "Total"` separated by spaces or tabs to form columns.

- Print a separator line using exactly 45 dashes (-).

- Print the user's inputted data under the respective headers using an f-string.

- `price` and `total_cost` must display exactly 2 decimal places.

- Print a bottom border using exactly 45 equal signs (=).

```python
item_name=input("Enter the item's name:")
price=float(input("Enter the price: "))
quantity=int(input("Enter the quantity: "))
total_cost=price*quantity
print("="*45)
print("Item Name\tPrice\tQty\tTotal")
print("-"*45)
print(f"{item_name:<15} {price:<6.2f}  {quantity:<6}  {total_cost:<5.2f}")
print("="*45)
```
