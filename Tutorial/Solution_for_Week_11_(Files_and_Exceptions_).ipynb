**Question 2**  
Manually create a file named `scores.txt` containing a list of students (one person per line: Name, Score).
```
An, 8
Binh, 4
Chau, 9
Dung, 3
```
Read the `scores.txt` file.  
Find the students with a score below 5, and write the names of those students into a new file named `failed.txt`.  
Read `failed.txt`

````python
with open("scores.txt",'w') as f:
  f.write("An, 8\n Binh, 4\n Chau, 9\n Dung, 3\n")
with open("scores.txt", 'r') as f, open('failed.txt','w') as fn:
  for line in f:
    name, score=line.strip().split(', ')
    if int(score)<5:
      fn.write(name + "\n")
with open('failed.txt', 'r') as fn:
  print(fn.read())

````

**Question 3**    
Write a program that continuously prompts the user to enter the name of a text file to read.

If the file does not exist and print a friendly message.

If the file exists, read the file.

Write the counted result into a file named `analysis_log.txt` using append mode to keep a history of the executions.

````python
while True:
    filename = input("Enter the name of a text file to read (or 'quit' to stop): ")

    if filename.lower() == 'quit':
        break

    try:
        with open(filename, 'r') as file:
            content = file.read()
            word_count = len(content.split())
            print(f"Successfully read {filename}. Total words: {word_count}")

        with open('analysis_log.txt', 'a') as log_file:
            log_file.write(f"File: {filename} - Word count: {word_count}\n")
        print("Result logged in analysis_log.txt\n")

    except FileNotFoundError:
        print(f"Sorry, the file '{filename}' does not exist. Please try again.\n")
````

**Question 4**

Write a function `count_words(filename)` that reads a text file and counts the approximate number of words using the `.split()` method.    
Return the count.    
If the file doesn't exist, return 0.
After that, test the function `count_words(filename)` with both cases.

````python
def count_words(filename):
    try:
        with open(filename, 'r') as file:
            content = file.read()
            words = content.split()
            return len(words)
    except FileNotFoundError:
        return 0

with open("test_words.txt", "w") as f:
    f.write("Hello world! This is a test file.")
#Test
print("Word count of existing file:", count_words("test_words.txt"))
print("Word count of non-existing file:", count_words("missing_file.txt"))
````

**Question 5**  
Write a Python program to manage a file named `daily_working_checkup.txt` with the following 2 features:
1. Add a log:

Prompt the user to input a task description. The program should automatically prepend the current time (down to the minute) to the input and append the complete line to the file.

2. Check & Statistics:  

Use a try-except block to open and read the file.  

If the file does not exist print a friendly message (e.g., "File does not exist yet.").

If the file exists, read its content and count how many tasks assigned.



````python
import datetime
def manage_daily_checkup():
    print("1. Add a log")
    print("2. Check & Statistics")
    choice = input("Choose a feature (1 or 2): ")

    if choice == '1':
        task = input("Enter a task description: ")
        current_time = datetime.datetime.now().strftime("%Y-%m-%d %H:%M")

        with open("daily_working_checkup.txt", "a") as file:
            file.write(f"[{current_time}] {task}\n")
        print("Task added successfully.")

    elif choice == '2':
        try:
            with open("daily_working_checkup.txt", "r") as file:
                tasks = file.readlines()
                task_count = len(tasks)
                print(f"Checkup complete. You have {task_count} assigned task(s).")
                for t in tasks:
                    print(t.strip())
        except FileNotFoundError:
            print("File does not exist yet. No tasks assigned.")
    else:
        print("Invalid choice.")

#Test program
manage_daily_checkup()
````

**Question 6**  
Write a program that stores user information (name, age, favorite programming language) in a dictionary.   

Save this dictionary into a file named `user_data.json`.

Write a startup function to read data from this file when the program launches.

Incorporate a try-except block to handle a FileNotFoundError: If the `user_data.json` file does not exist yet, automatically trigger a function to prompt the user for their information and create a new file.

````python
import json

def get_user_input():
    print("File not found! Let's create your profile.")
    name = input("Enter your name: ")
    age = int(input("Enter your age: "))
    language = input("Enter your favorite programming language: ")

    data = {
        "name": name,
        "age": age,
        "language": language
    }

    with open("user_data.json", "w") as file:
        json.dump(data, file)
    print("Data saved successfully.")
    return data

def startup():
    try:
        with open("user_data.json", "r") as file:
            data = json.load(file)
            print("Welcome back!")
            print(f"Name: {data['name']}, Age: {data['age']}, Favorite Language: {data['language']}")
    except FileNotFoundError:
        get_user_input()
#Test
startup()
````

**Question 7**

Create a dictionary of 2 trivia questions and answers.   
Ask user's name to distiguish players.   
Loop through questions and ask the user.  
Calculate their score (+1 with 1 correct answer, no deduction with wrong answers).    
Using `json.load()` and `json.dump()`, maintain a high-score list in `scores.json`.   
Append the current user's name and score to the JSON file without overwriting previous players' scores.

````python
import json

def trivia_game():
    trivia = {
        "What is the capital of Japan?": "Tokyo",
        "What is 8 + 5?": "13"
    }
    player_name = input("Enter your name: ")
    score = 0
    for question, correct_answer in trivia.items():
        print(question)
        answer = input("Your answer: ")
        if answer.lower().strip() == correct_answer.lower():
            print("Correct!\n")
            score += 1
        else:
            print(f"Wrong! The correct answer is {correct_answer}.\n")

    print(f"Game over, {player_name}! Your score is {score}/{len(trivia)}.")
    filename = "scores.json"
    scores_list = []

    try:
        with open(filename, "r") as file:
            scores_list = json.load(file)
    except FileNotFoundError:
        scores_list = []

    scores_list.append({
        "name": player_name,
        "score": score
    })
    with open(filename, "w") as file:
        json.dump(scores_list, file, indent=4)

    print("Your score has been saved to scores.json.")


trivia_game()
````
