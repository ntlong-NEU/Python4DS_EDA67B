### 1. Essential Python String Methods
| Method | Description | Example |
| :--- | :--- | :--- |
| `s.lower()` / `s.upper()` | Converts string to lowercase / uppercase | `"PyThOn".lower()` $\rightarrow$ `'python'` |
| `s.strip()` | Removes leading and trailing whitespace | `"  hello  ".strip()` $\rightarrow$ `'hello'` |
| `s.replace(old, new)` | Replaces occurrences of a substring | `"a-b-c".replace("-", " ")` $\rightarrow$ `'a b c'` |
| `s.split(sep)` | Splits string into a list of words | `"a b c".split()` $\rightarrow$ `['a', 'b', 'c']` |
| `" ".join(list_words)` | Concatenates list elements with a delimiter | `"-".join(['a', 'b'])` $\rightarrow$ `'a-b'` |
| `s.count(sub)` | Counts occurrences of a substring | `"banana".count("a")` $\rightarrow$ `3` |
| `len(s)` | Returns the length of the string | `len("hello")` $\rightarrow$ `5` |

---

### 2. Regular Expression Metacharacters (`import re`)
| Pattern | Meaning | Example Match |
| :---: | :--- | :--- |
| `\d` | Any digit `[0-9]` | `'7'` in `"agent 007"` |
| `\D` | Any non-digit character `[^0-9]` | `'a'` in `"abc123"` |
| `\w` | Any alphanumeric character or underscore `[a-zA-Z0-9_]` | `'user_01'` in `"@user_01"` |
| `\W` | Any non-alphanumeric character | `'!'`, `'@'`, `' '` |
| `\s` | Any whitespace (space, tab, newline `[ \t\n\r]`) | `' '` or `'\n'` |
| `\S` | Any non-whitespace character | Any visible character |
| `\b` | Word boundary | `\bcat\b` matches `"cat"`, but not `"scatter"` |
| `.` | Any character except a newline | `a.c` matches `"abc"`, `"a1c"` |
| `^` | Start of a string | `^Hello` matches `"Hello world"` |
| `$` | End of a string | `end$` matches `"the end"` |

---

### 3. Quantifiers & Groups
| Pattern | Meaning | Example |
| :---: | :--- | :--- |
| `+` | 1 or more repetitions | `\d+` matches `"123"` |
| `*` | 0 or more repetitions | `ab*c` matches `"ac"`, `"abc"`, `"abbc"` |
| `?` | 0 or 1 repetition (optional) | `https?` matches `"http"` and `"https"` |
| `{n}` | Exactly $n$ times | `\d{4}` matches `"2026"` |
| `{n,}` | At least $n$ times | `\d{2,}` matches `"12"`, `"12345"` |
| `{n,m}` | Between $n$ and $m$ times | `\d{2,4}` matches 2 to 4 digits |
| `[...]` | Character set (any single char inside) | `[aeiou]` matches any vowel |
| `[^...]` | Negated set (any char NOT inside) | `[^0-9]` matches any non-digit |
| `(...)` | Capture group (extracts sub-matches) | `(\w+)@(\w+\.\w+)` |
| `a|b` | Alternation (matches `a` or `b`) | `cat|dog` matches `"cat"` or `"dog"` |

---

### 4. Key `re` Module Functions
- `re.search(pattern, text)`: Searches for the **first** match anywhere in the text. Returns a `Match` object if found, or `None`.
- `re.findall(pattern, text)`: Returns a **list of all matches**. If capture groups `()` are used, returns a list of tuples.
- `re.sub(pattern, replacement, text)`: Replaces all matching occurrences with a new string.
