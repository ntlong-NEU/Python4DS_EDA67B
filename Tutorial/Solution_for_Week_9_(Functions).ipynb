**Question 1**

Write a function named calculate_coffee_bill(cups, price_per_cup) that takes 2 parameters:
1. cups: The number of coffee cups purchased (an integer).
2. price_per_cup: The cost of a single cup of coffee (an integer or float).

If a customer buys 5 or more cups, the shop gives a 10% discount on the total bill.


````python
def calculate_coffee_bill(cups, price_per_cup):
    total_bill = cups * price_per_cup
    if cups >= 5:
        total_bill *= 0.9
    return total_bill
print(calculate_coffee_bill(4, 3.0))
print(calculate_coffee_bill(5, 3.0))
````

**Question 2**. Prime Number Check and Sum

- Write a function is_prime(n) that takes an integer n as input and returns True if n is a prime number, and False otherwise.

- Write a program using the above function to calculate the sum of all prime numbers from 0 to N (where N is entered from the keyboard).

````python
def is_prime(n):
    if n < 2:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True
N = int(input("Enter N: "))
total_sum=0
for i in range(0,N+1):
  if is_prime(i):
    total_sum+=i
print(total_sum)
````

**Question 4**

Write a function `launch_mission` that accepts a mission name (required), a destination (defaulting to "Mars"), an arbitrary number of crew members, and an arbitrary number of vessel specifications as keyword arguments.   

The function must output the exact format below when executed with `the mission "Apollo X", destination "Europa", crew members "Neil", "Buzz", "Sally"`, and vessel specifications `fuel_type="Antimatter", engine_count=4, and shield_status="Active"`:

```
Launching Mission: Apollo X
Destination: Europa
Crew Members:
  - Neil
  - Buzz
  - Sally
Vessel Specifications:
  - Fuel_Type: Antimatter
  - Engine_Count: 4
  - Shield_Status: Active
```

````python
def launch_mission(mission_name, destination="Mars", *crew, **vessel_specs):
    print(f"Launching Mission: {mission_name}")
    print(f"Destination: {destination}")
    print("Crew Members:")
    for member in crew:
        print(f"  - {member}")
    print("Vessel Specifications:")
    for spec, val in vessel_specs.items():
        print(f"  - {spec.title().replace('_',' ')}: {val}")
launch_mission(
    "Apollo X",
    "Europa",
    "Neil", "Buzz", "Sally",
    fuel_type="Antimatter",
    engine_count=4,
    shield_status="Active"
)
````

**Question 5**

 Write a generator function `infinite_multiples(n, max_limit=None)` that takes a number n. It should yield multiples of n (e.g., n × 1, n × 2, n × 3...) indefinitely. If a max_limit is provided, stop yielding when the multiple exceeds that limit.

````python
def infinite_multiples(n, max_limit=None):
    counter = 1
    while True:
        multiple = n * counter
        if max_limit is not None and multiple > max_limit:
            break
        yield multiple
        counter += 1
print(list(infinite_multiples(3, max_limit=15)))
````
